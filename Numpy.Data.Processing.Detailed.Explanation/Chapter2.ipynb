#### 编写同样的代码，对ndarray与Python的列表的性能进行比较。

```python
def np_append():
    a = np.array([1, 2, 3])
    for i in range(10000):
        a = np.append(a, [i])
    return a
```

```python
def list_append():
    a = [1, 2, 3]
    for i in range(10000):
        a.append(i)
    return np.array(a)
```

```python
%timeit np_append()
```

```python
%timeit list_append()
```

## 2.4 指定条件获取元素索引的函数

### 2.4.1 np.where

```python
import numpy as np
```

```python
a = np.array([10, 12, 9, 3, 19])
```

```python
a[a<10]
```

```python
a = np.arange(20, 0, -2) # 首先创建一维数组
```

```python
a
```

```python
np.where(a < 10) #  获取小于10的索引
```

```python
a[np.where(a < 10)]
```

```python
a = np.arange(12).reshape((3, 4))  # 指定为3×4的二维数组
```

```python
a
```

```python
np.where(a%2==0) #  只取出偶数元素
```

```python
np.where(a%2==0, 'even', 'odd') #  偶数返回even，奇数返回odd
```

```python
np.where(a%2==0, 'even') # 如果只设置True，就会导致运行时产生错误
```

```python
np.where(a%2==0, 'even', 'odd') # 偶数返回even，奇数返回odd
```

```python
b = np.reshape(a, (3, 4))
```

```python
c = b ** 2
```

```python
c
```

```python
np.where(b%2==0, b, c) # 只有奇数元素被转换成了c中的元素
```

```python
np.where(b%2==0, b, (10, 8, 6, 4)) #  运用广播机制，使用(10, 8, 6, 4)重复的值
```

### 2.6.4 np.argmin

```python
d = np.array([
[1.2, 1.5, 2.3, 1.8],
[0.2, 2.5, 2.1, 2.0],
[3.1, 3.3, 1.5, 2.1]])
```

```python
d.argmin() # 首先在不指定参数的情况下执行代码
```

```python
np.argmin(d) # 同样地执行代码
```

```python
np.unravel_index(np.argmin(d), d.shape) # 这样调用，返回的就是没有被降为一维数组的索引
```

```python
np.argmin(d, axis=0) # 接着对坐标轴进行指定
```

```python
np.argmin(d, axis=1)
```

```python
d.argmin(axis=1) # ndarray.argmin也可以完成同样的处理
```
