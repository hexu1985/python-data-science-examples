```python
import math
g = 9.81  # m/s^2
m = 68.1  # kg
t = 12    # s
cd = 0.25 # kg/m
v = math.sqrt(m*g/cd)*math.tanh(math.sqrt(cd*g/m)*t)
print('velocity = ', v, ' m/s')
```

```python
import math
import scipy.constants as pc
def freefall(t, m, cd):
    g = pc.g
    return math.sqrt(m*g/cd)*math.tanh(math.sqrt(cd*g/m)*t)
```

```python
freefall(12, 68.1, 0.25)
```
