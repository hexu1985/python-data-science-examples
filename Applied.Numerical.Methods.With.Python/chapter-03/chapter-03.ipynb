```python
import math
import scipy.constants as pc
def freefall(t, m, cd):
    """
    function freefall: computes bungee jumper velocity
            with second-order drag
    input arguments:
    t = time (s), m = mass (kg), cd = drag coefficient (kg/m)
    returns velocity in m/s
    """
    g = pc.g
    return math.sqrt(m*g/cd)*math.tanh(math.sqrt(cd*g/m)*t)
```

```python
freefall(12, 68.1, 0.25)
```

```python
'velocity = {0:g} m/s'.format(freefall(8, 100, 0.25))
```

```python
import numpy as np
def stats(x):
    n = len(x)
    avg = np.average(x)
    s = np.std(x)
    return n, avg, s
```

```python
y = [8,5,10,12,6,7,5,4]
stats(y)
```

```python
import numpy as np
def sgnsqr(x):
    x1 = x*abs(x)
    return x1
    
def sgnsqrt(x):
    # print('x1 = ', x1)
    x2 = np.sqrt(abs(x))*np.sign(x)
    return x2

x = -2.0
print('x squared with sign = ', sgnsqr(x))
print('square root of x with sign = ', sgnsqrt(x))
#print('x2 = ', x2)
```

```python
def tempcovert(tempin, units='degF'):
    if units == 'degF':
        return (tempin-32)/1.8
    else:
        return tempin*1.8+32    
```
