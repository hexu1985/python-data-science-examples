### 3.1.2 Python函数

```python
import math
import scipy.constants as pc
def freefall(t, m, cd):
    """
    function freefall: computes bungee jumper velocity
            with second-order drag
    input arguments:
    t = time (s), m = mass (kg), cd = drag coefficient (kg/m)
    returns velocity in m/s
    """
    g = pc.g
    return math.sqrt(m*g/cd)*math.tanh(math.sqrt(cd*g/m)*t)
```

```python
freefall(12, 68.1, 0.25)
```

```python
'velocity = {0:g} m/s'.format(freefall(8, 100, 0.25))
```

## 3.4 嵌套和缩进

```python
def quadroots(a,b,c):
    import math
    if a == 0:
        if b != 0:
            r1 = -c/b
            return r1
        else:
            print('Trivial solution. Try again.')
    else:
        d = b**2 - 4*a*c
        if d >= 0:
            r1 = (-b+math.sqrt(d))/(2*a)
            r2 = (-b-math.sqrt(d))/(2*a)
            i1 = 0
            i2 = 0
        else:
            r1 = -b/(2*a)
            i1 = math.sqrt(abs(d))/(2*a)
            r2 = r1
            i2 = -i1
        return r1,i1,r2,i2
```

```python
quadroots(1,1,1)
```

```python
quadroots(1,5,1)
```

```python
quadroots(0,5,1)
```

```python
quadroots(0,0,0)
```

### 3.5.2 函数-函数

```python
import numpy as np
def midpoint(f,x1,x2):
    xmid = (x1+x2)/2
    return (f(xmid))

def f(w):
    return np.sin(w)*np.cosh(w)-5

fmid = midpoint(f,0,3)
print('function value at midpoint =',fmid)
```

```python
g = lambda x: np.pi*x**2/4
midpoint(g,1.5,2.78)
```
